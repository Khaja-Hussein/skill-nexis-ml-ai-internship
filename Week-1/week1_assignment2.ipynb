## Assignment 2: Handling Missing Data

```python
import pandas as pd
```

```python
data = {
    "Name": ["Ali", "Rahul", "Sara", "John", "Aisha"],
    "Age": [20, 21, None, 22, None],
    "Marks": [85, None, 90, 78, 88]
}

df = pd.DataFrame(data)

df
```

```python
df.isnull().sum()
```

```python
age_mean = df["Age"].mean()
print(age_mean)
```

```python
df["Age"] = df["Age"].fillna(age_mean)
df
```

```python
marks_median = df["Marks"].median()
print(marks_median)
```

```python
df["Marks"] = df["Marks"].fillna(marks_median)
df
```

```python
df.isnull().sum()
```
