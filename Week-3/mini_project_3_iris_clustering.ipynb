# Mini Project 3: Iris Flower Clustering

## K-Means Clustering on Iris Dataset

### Objective
Apply K-Means clustering to the Iris dataset using 3 clusters, visualize the clusters, and compare the predicted clusters with the true labels.

```python
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.datasets import load_iris
from sklearn.cluster import KMeans
```

```python
iris = load_iris()

X = iris.data
y = iris.target

print("Dataset shape:", X.shape)
print("Feature names:", iris.feature_names)
```

```python
kmeans = KMeans(
    n_clusters=3,
    random_state=42,
    n_init=10
)
```

```python
kmeans.fit(X)
```

```python
predicted_clusters = kmeans.labels_

print("Predicted clusters:")
print(predicted_clusters)
```

```python
plt.figure(figsize=(8, 5))

plt.scatter(
    X[:, 0],
    X[:, 1],
    c=predicted_clusters,
    cmap="viridis",
    s=50
)

plt.xlabel("Sepal Length (cm)")
plt.ylabel("Sepal Width (cm)")
plt.title("Iris Flower Clustering using K-Means")

plt.show()
```

```python
print("True label counts:")
print(pd.Series(y).value_counts().sort_index())

print("\nPredicted cluster counts:")
print(pd.Series(predicted_clusters).value_counts().sort_index())
```

```python
comparison = pd.crosstab(
    pd.Series(y, name="True Label"),
    pd.Series(predicted_clusters, name="Predicted Cluster")
)

print(comparison)
```

## Conclusion

K-Means clustering was applied to the Iris dataset using 3 clusters.

The clusters were visualized using a scatter plot and compared with the true Iris labels using a cross-tabulation.

The results show that the predicted clusters have a strong relationship with the true labels, although some samples from the second and third classes were grouped into the same cluster.

K-Means cluster numbers are arbitrary and do not directly represent the original class labels.
