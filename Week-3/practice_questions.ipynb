## Clustering & Dimensionality Reduction

### Question 1: Apply K-Means on Iris Dataset and Print Cluster Centers

```python
from sklearn.datasets import load_iris
from sklearn.cluster import KMeans
```

```python
iris = load_iris()

X = iris.data

print("Dataset shape:", X.shape)
```

```python
kmeans = KMeans(
    n_clusters=3,
    random_state=42,
    n_init=10
)

kmeans.fit(X)

print("Cluster Centers:")
print(kmeans.cluster_centers_)
```

### Question 2: Visualize the Clusters using Matplotlib

```python
import matplotlib.pyplot as plt

clusters = kmeans.labels_

plt.figure(figsize=(8, 5))

plt.scatter(
    X[:, 0],
    X[:, 1],
    c=clusters,
    cmap="viridis",
    s=50
)

plt.xlabel("Sepal Length (cm)")
plt.ylabel("Sepal Width (cm)")
plt.title("K-Means Clusters on Iris Dataset")
plt.show()
```

### Question 3: Apply PCA to Reduce the Dataset to 2D

```python
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
```

```python
scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)

pca = PCA(n_components=2)

X_pca = pca.fit_transform(X_scaled)

print("Original shape:", X.shape)
print("PCA shape:", X_pca.shape)
```

### Question 4: Find Explained Variance Ratio

```python
print("Explained Variance Ratio:")
print(pca.explained_variance_ratio_)

print("Total Explained Variance:")
print(pca.explained_variance_ratio_.sum())
```

## Conclusion

K-Means clustering and PCA were applied to the Iris dataset.

K-Means successfully grouped the data into 3 clusters and the clusters were visualized using a scatter plot.

PCA reduced the dataset from 4 dimensions to 2 dimensions. The first two principal components explain approximately 95.81% of the total variance.
